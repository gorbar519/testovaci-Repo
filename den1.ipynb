```python
def find_short_words (l:list):
  list_new = []
  for a in l:
    if  len(a)< 5:
        list_new.append(a)
  return list_new

l = find_short_words(['the', 'scariest', 'moment', 'is', 'always', 'just', 'before', 'you', 'start'])
print(l);

def find_short_words1(l:list):
 return  [a for a in l if len(a) < 5]


g = find_short_words1(['the', 'scariest', 'moment', 'is', 'always', 'just', 'before', 'you', 'start'])
print(g);
```

```python
# {"first_name": "Jane", "last_name": "Kowalski", "full_name": "Jane Kowalski", "initials": "J. K."}

def user_data(first_name:str, last_name:str):
  d = {}
  d['first_name'] = first_name
  d['last_name'] = last_name
  d['full_name'] = first_name + ' ' + last_name
  d['initials'] = first_name[0] + '. ' + last_name[0] + '.'
  return d

print(user_data('Jane', 'Kowalski'))

```

```python
def display_dict(dictionary:dict):
  for d in dictionary:
    print(d +'=>'+ dictionary[d])

display_dict({'cat': 'Kot', 'dog': 'Pies', 'bird': 'Ptak'})

#v2
def display_dict1(dictionary:dict):
    [print(d +'=>'+ dictionary[d]) for d in dictionary]

display_dict1({'cat': 'Kot', 'dog': 'Pies', 'bird': 'Ptak'})

```

```python
def wordfinder(list_of_words, word):
  if word in list_of_words:
    list_of_words.append("Found")


words = ['Twinkle', 'twinkle', 'little', 'star']
print('words BEFORE:', words)
wordfinder(words, 'twinkle')
print('words AFTER:', words)

```

```python
def create_email(first_name, last_name):
  return f"{first_name[0].lower()}.{last_name.lower()}@myamazingcompany.com"

print(create_email('John', 'Connor'))
```

```python
#Hand-deliver to: John Connor
#Hayvenhurst Drive Apt. 12
#14329 Van Nyus

def create_address(first_name, last_name, street, number, city, zipcode):
  return f"Hand-deliver to: {first_name} {last_name}\n{street} Apt. {number}\n{zipcode} {city}"

print(create_address('John', 'Connor', 'Hayvenhurst Drive', 12, 'Van Nyus', '14329'))
```

```python
def anonymize(email):
  if len(email) >= 10:
    return f"{email[:3]}***{email[-5:]}"
  else:
    return f"***{email[-5:]}"

print(anonymize('j.connor@gmail.com'))
print(anonymize('jco@o2.pl'))
```

```python
def anonymize(email):
  return f"***{email[email.find('@'):]}"

print(anonymize('j.connor@gmail.com'))
print(anonymize('jco@o2.pl'))
print(anonymize('jcoo2.pl'))
```
